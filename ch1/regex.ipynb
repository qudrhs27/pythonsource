#### 정규표현식(정규식)
- 복잡한 문자열을 처리할 때 사용하는 방법
- 파이썬만의 고유한 문법이 아니라 문자열을 처리하는 모든 곳에서 사용하는 일종의 형식 언어

```python
# 주민등록번호 -> 뒷자리를 모두 * 문자로 변경

# 파이썬에서 정규식 지원하는 모듈
import re

dir(re)

# re.compile() : 정규식을 작성

pattern = re.compile("ab*")

# 패턴을 특정 문자열이 포함하고 있는가?
print(pattern.match("abc"))
```

| 메서드 | 목적 |
| ---- | ---- |
| match | 문자열의 처음부터 정규식과 매치되는지 조사 |
| search | 문자열 전체를 검색하여 정규식과 매치되는지 조사 |
| findall | 정규식과 매치되는 모든 문자열(substring)을 리스트로 리턴 |
| finditer | 정규식과 매치되는 모든 문자열(substring)을 이터레이더 객체로 리턴 |

```python
# [문자] : [] 안에 들어있는 문자들과 매치하느냐?

p = re.compile("[abc]")

print(p.match("a"))
print(p.match("before"))
print(p.match("dude"))
```

```python
m = p.match("a")

print(f"패턴 시작 : {m.start()}")
print(f"패턴 끝 : {m.end()}")
print(f"패턴과 일치하는 문자열 반환 : {m.group()}")
print(f"일치의 시작,끝 반환 : {m.span()}")

if m:
    print(f"Match found {m.group()}")
else:
    print("No Match")
```

- 자주 사용하는 문자 표현식

| 정규표현식 | 설명 |
| ---- | ---- |
| \d | 숫자와 매치 [0-9]와 동일한 표현식 |
| \D | 숫자가 아닌 것과 매치 [^0-9]와 동일한 표현식 |
| \s | 화이트스페이스(스페이스,탭,줄바꿈 등) 문자와 매치 [빈칸\t\n\r\f\v] 와 동일한 표현식 |
| \S | 화이트스페이스(스페이스,탭,줄바꿈 등) 문자가 아닌 것과 매치 [^빈칸\t\n\r\f\v] 와 동일한 표현식 |
| \w | 문자 + 숫자와 매치 [a-zA-Z0-9_] 와 동일한 표현식 |
| \W | 문자 + 숫자가 아닌 것과 매치 [^a-zA-Z0-9_] 와 동일한 표현식 |

```python
p = re.compile("[0-9]+")
p = re.compile(r"\d+")
print(p.match("012 abc"))

# 원본 문자열에서 0~9가 아닌
# p = re.compile("[^0-9]+")
p = re.compile(r"\D+")
print(p.match("012 abc"))

# 공백,탭,엔터 문자 찾기
p = re.compile("[\t\n\r\f\v]+")
p = re.compile(r"\s+")
print(p.search("012 abc"))

# 공백,탭,엔터 문자가 아닌 문자 찾기
# p = re.compile("[^ \t\n\r\f\v]+")
p = re.compile(r"\S+")
print(p.search("012 abc"))

# 대소문자,숫자,_ 포함한 문자
# p = re.compile("[a-zA-Z0-9_]+")
p = re.compile(r"\w+")
print(p.search("012 abc"))

# 대소문자,숫자,_ 가 아닌 문자 찾기
# p = re.compile("[^a-zA-Z0-9_]+")
p = re.compile(r"\W+")
print(p.search("012 abc"))
```

```python
# re.I : ignorecase

# 영어가 포함되어 있는지?
p = re.compile("[a-z]+", re.I)
print(p.match("python"))
print(p.match("Python"))
print(p.match("PYTHON"))
```

```python
# re.M : MULTILINE

data = """python one
life is too short
python two
you need python
python three
"""

# ^ : 시작하는
# python 시작하고 스페이스문자하나 영어,숫자,_ 인 문자를 찾아줘
p = re.compile(r"^python\s\w+", re.M)
print(p.findall(data))
```

```python
# | : or

p = re.compile("Crow|Servo")
m = p.match("CrowHello")
print(m)

m = p.match("ServoHello")
print(m)
```

```python
# ^ : 시작하는,   $ : 끝나는

p = re.compile("^Life")
m = p.search("Life is too short")
print(m)

m = p.search("My Life")
print(m)

p = re.compile("short$")
m = p.search("Life is too short")
print(m)
```

```python
# () : 그루핑

# ABCABC 문자열이 반복되는지 조사

p = re.compile("[ABC]+")
m = p.search("ABCABCABC OK?")
print(m)
print(m.group())
```

```python
p = re.compile("(ABC)+")
m = p.search("ABCABCABC OK?")
print(m)
print(m.group())
print(m.group(1))
```

```python
# 이름 010-1234-5678
p = re.compile(r"(\w+)\s+(\d{3}-\d{4}-\d{4})")
m = p.search("park 010-1234-5678")
print(m)
print(m.group())
print(m.group(1))
print(m.group(2))
```

```python
# 찾아서 바꾸기
# John Smith
# Smith, John

p = re.compile(r"(\w+)\s(\w+)")
m = p.search("John Smith")
print(m)

# re.sub(패턴, 바꿀 문자열, 원본 문자열)
re.sub(p, r"\2, \1", "John Smith")
```

```python
# re.compile()
# re.sub()

# re.fullmatch() : 문자열 전체가 패턴과 일치

p = re.compile(r"\d+")

print(re.fullmatch(p, "1232abc"))
print(re.fullmatch(p, "123456"))
print(re.fullmatch(p, "abc123"))
```

```python
# 영문 소문자로 시작하고 영문,숫자 포함된 6~12글자

p = re.compile("[a-z][a-z0-9]{5,11}")
print(re.fullmatch(p, "1232abc"))
print(re.fullmatch(p, "123456"))
print(re.fullmatch(p, "abc123"))

m = re.fullmatch(p, "abc123")
print(m.group())
```

```python
# 아이디,비밀번호,전화번호 입력 받기
# 패턴에 맞을 때까지 입력받기

# 아이디 검사 조건
# 영문 대문자로 시작, 영문,숫자,특수문자(!@#$&) 포함, 6~15자

# 비밀번호 검사 조건
# 영문 대소문자,숫자,특수문자(!@#$%) 포함, 8자 이상

# 전화번호 010-1234-5678 / 01012345678
# fullmatch()
import re

p_id = re.compile("[A-Z][a-zA-Z0-9!@#$&]{5,14}")
while True:
    user_id = input("아이디를 입력하세요 : ")
    if re.fullmatch(p_id, user_id):
        print("아이디 입력 완료")
        break
    else:
        print("아이디를 다시 입력하세요.")

p_pwd = re.compile("[a-zA-Z0-9!@#$%]{8,}")
while True:
    user_pwd = input("비밀번호를 입력하세요 : ")
    # if re.fullmatch(p_pwd, user_pwd):
    if (len(user_pwd) >= 8
    and re.search("[A-Z]", user_pwd)
    and re.search("[a-z]", user_pwd)
    and re.search(r"\d", user_pwd)
    and re.search("[!@#$%&]", user_pwd)
    ):
        print("비밀번호 입력 완료")
        break
    else:
        print("비밀번호를 다시 입력하세요.")

p_tel = re.compile(r"^\d{3}-?\d{4}-?\d{4}$")
while True:
    user_tel = input("전화번호를 입력하세요 : ")
    if re.fullmatch(p_tel, user_tel):
        print("전화번호 입력 완료")
        break
    else:
        print("전화번호를 다시 입력하세요.")

print(f"아이디 : {user_id}\n비밀번호 : {user_pwd}\n전화번호 : {user_tel}")
```

```python
# # 로그 파일에서 필요한 정보만 추출하기
# # ERROR 만 출력
# # INFO 개수, ERROR 개수 출력
# # 날짜별 ERROR 발생 횟수 출력

# # 출력예시
# ERROR 목록
# ---------------
# Database Connection Failed
# Invalid Password

# INFO : 3
# ERROR : 2

# 2026-07-20 : 1
# 2026-07-21 : 1


with open("./data/log.txt", "r", encoding="utf-8") as f:
    logs = f.read()

# enter 기준으로 분리
logs = logs.splitlines()

err_logs, info = [], 0
level = re.compile("ERROR|INFO")
for log in logs:
    if re.search(level, log):
        if "INFO" in log:
            info += 1
        else:
            err_logs.append(log)

print(err_logs)

print("ERROR 목록")
print("-" * 20)

daily_error = {}
for log in err_logs:
    error_split = log.split("ERROR")
    print(error_split[1].strip())
    # if error_split[0] in daily_error:
    #     daily_error[error_split[0]] += 1
    # else:
    #     daily_error[error_split[0]] = 1

    daily_error[error_split[0]] = daily_error.get(error_split[0], 0) + 1

print(f"INFO : {info}")
print(f"ERROR : {len(err_logs)}")

for key, value in daily_error.items():
    print(f"{key} : {value}")
```

```python
# (?P<이름>패턴) : 그루핑된 문자열에 이름 붙이기

p = re.compile(r"(?P<name>\w+)\s+(?P<tel>\d{3}-\d{4}-\d{4})")
m = p.search("park 010-1234-5678")
print(m)
print(m.group())
print(m.group("name"))
print(m.group("tel"))
```

```python
# the the 추출
# 정규식 안에서 재참조 가능
p = re.compile(r"(?P<word>\b\w+)\s+(?P=word)")
m = p.search("Paris in the the spring")
print(m)
```

```python
error, info = 0, 0
daily_error = {}
pattern = re.compile(r"(?P<date>\d{4}-\d{2}-\d{2})\s(?P<level>INFO|ERROR)\s(?P<message>.+)")

for log in logs:
    m = pattern.search(log)

    if not m:
        continue

    date = m.group('date')
    level = m.group('level')
    message = m.group('message')

    if level == "INFO":
        info += 1
    elif level == "ERROR":
        error += 1
        print(message)
        daily_error[date] = daily_error.get(date, 0) + 1


print("ERROR 목록")
print("-" * 20)

print(f"INFO : {info}")
print(f"ERROR : {len(err_logs)}")

for key, value in daily_error.items():
    print(f"{key} : {value}")
```

```python
# 전방탐색

url = "http://google.com"

# 프로토콜 검색 ftp://~~

p = re.compile(".+:")
m = p.search(url)
print(m)
print(m.group())
# : 패턴으로는 쓰고 매칭되는 문자열 돌려받을 때는 빼고 받고 싶다면?
```

```python
# ?= : 긍정형 전방탐색
# ?! : 부정형 전방탐색
p = re.compile(".+(?=:)")
m = p.search(url)
print(m.group())
```

```python
# 영문대소문자, 숫자, 특수문자 포함, 8자 이상
pwd = re.compile(r"^(?=.*[A-Z])(?=.*[a-z])(?=.*\d)(?=.*[!@#$%&])[A-Za-zA-Z0-9!@#$&]{8,}$")

print(pwd.fullmatch("A1bc0101010"))
print(pwd.fullmatch("abc010#1010"))
print(pwd.fullmatch("12345678"))
```

```python
# greedy 와 non-greedy

s = "<html><head><title>Title</title></head></html>"

# greedy
p = re.compile("<.+>")
print(p.search(s).group())

# non-greedy
# *?, {m,n}?
p = re.compile("<.+?>")
print(p.search(s).group())
```
